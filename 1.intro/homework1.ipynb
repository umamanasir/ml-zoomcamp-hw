```python
list = ['vehicle_weight', 'model_year']
x = df[df['origin'] == 'Asia'][list].head(7)
```

```python
x
```

```python
transpose = x.T
transpose
```

```python
XTX = transpose.dot(x)
XTX
```

```python
inv_XTX = np.linalg.inv(XTX)
```

```python
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
```

```python
res = inv_XTX.dot(transpose)
w = res.dot(y)
w
```
